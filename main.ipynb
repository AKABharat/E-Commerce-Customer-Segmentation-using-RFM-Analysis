**Step 3**

- calculating 
  - R_Score → 1 to 4 (4 = most recent buyer)
  - F_Score → 1 to 4 (4 = most frequent buyer)
  - M_Score → 1 to 4 (4 = highest spender)

- Pandas "qcut"
  - The pandas.qcut function is a quantile-based discretization tool that splits data into equal-sized buckets based on rank or sample quantiles.
  - ```pandas.qcut(x, q, labels=None, retbins=False, precision=3, duplicates='raise') ```
  
  Parameters:-
  - x: The input 1D array or pandas Series to be binned.q: The binning criteria.Integer: e.g., 4 for quartiles, 10 for deciles.
  - List of floats: Explicit quantiles, e.g., [0, .25, .5, .75, 1.].
  - labels (Optional): custom names for the resulting bins (e.g., ['Low', 'Medium', 'High']). 
    - If False, it returns only integer indicators.
    - retbins (Optional): If True, returns both the binned data and the calculated bin edges.
    - precision: Sets the number of decimal places for bin labels.
    - duplicates: If bin edges are not unique, set to 'drop' to remove them or 'raise' (default) to get a ValueError.

````python
# R_score
rfm_df["R_score"] = pd.qcut(rfm_df["Recency"],4, labels=[4,3,2,1])
````

````python
rfm_df
````

````python
rfm_df.describe()
````

- Based on this data, here are sensible boundaries:
- ScoreFrequency RangeBusiness Meaning11 purchaseOne time buyers22 to 3 purchasesOccasional buyers34 to 6 purchasesRegular buyers47 to 248 purchasesLoyal buyers
- So your bin edges should be:

- [0, 1, 3, 6, 248]


- 0 to 1 → Score 1
- 1 to 3 → Score 2
- 3 to 6 → Score 3
- 6 to 248 → Score 4

````python
rfm_df["F_score"] = pd.cut(rfm_df["Frequency"], bins=[0,1,3,5,248], labels=[1,2,3,4])
````

````python
# M_score
rfm_df["M_score"] = pd.qcut(rfm_df["Monetary"], 4, labels=[1,2,3,4], duplicates="drop");
````

````python
rfm_df
````

---
STEP 4: Create RFM_Score Column

````python
rfm_df["RFM_Score"] = rfm_df["R_score"].astype('str') + rfm_df["F_score"].astype('str') + rfm_df["M_score"].astype('str')
````
