**Guide from Step 2**

**Calcuating Recency** = (Latest Date from Dataset + 1) - Last Trasation Date [OF EACH CUSTOMER]
- snapshot_date = Latest Date from Dataset

```python
max_date = df['InvoiceDate'].max()
max_date
```

```python
snapshot_date = max_date + timedelta(days=1) # adding one day
snapshot_date
```

```python
latest_InvoiceDate = cleared_df.groupby("CustomerID")["InvoiceDate"].max()
latest_InvoiceDate
```

```python
cleared_df["CustomerID"].unique()
```

```python
rfm_df = pd.DataFrame(cleared_df["CustomerID"].unique()) # dataframe for rfm analysis [unique customer id only and no duplicate]
```

```python
rfm_df.columns = ["CustomerID"]
```

```python
rfm_df
```

```python
pd.to_datetime(latest_InvoiceDate)
```

```python
recency_series = (pd.to_datetime(snapshot_date) - latest_InvoiceDate).dt.days
```

```python
rfm_df["Recency"] = rfm_df["CustomerID"].map(recency_series)
```

```python
rfm_df
```

- The Core Difference
  - datetime (or dt in Pandas): Represents a specific point in time (e.g., January 1st, 2024, at 10:00 AM).
  - timedelta: Represents a duration or length of time (e.g., 5 days, 3 hours, or 20 minutes). 


---
**Calculating Frequency** = total count of unique invoice of each customer

```python
rfm_df
```

```python
frequency_series= cleared_df.groupby("CustomerID")["InvoiceNo"].nunique()
```

```python
# we want unique invoice count 
rfm_df["Frequency"] = rfm_df["CustomerID"].map(frequency_series)
```

```python
rfm_df
```

---

**Calculating Monetary** = sum of total "TotalPrice" of each customer group by id

```python
monetary_series = cleared_df.groupby("CustomerID")["TotalPrice"].sum();
```

```python
rfm_df["Monetary"] = rfm_df["CustomerID"].map(monetary_series);
```
